```python
import json
import random
import numpy as np
import torch

from pathlib import Path
```

```python
# Define project directory
PROJECT_DIR = Path.cwd()

if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

# Define project folders
DATA_DIR = PROJECT_DIR / "data"
CHECKPOINT_DIR = PROJECT_DIR / "checkpoints"
RESULTS_DIR = PROJECT_DIR / "results"

# Create directories
for directory in [DATA_DIR, CHECKPOINT_DIR, RESULTS_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

print("Project directory:", PROJECT_DIR)
```

```python
# Load project configuration
CONFIG_PATH = PROJECT_DIR / "config.json"

with open(CONFIG_PATH, "r") as file:
    CONFIG = json.load(file)

GENERAL_CONFIG = CONFIG["general"]
MODEL_CONFIGS = CONFIG["models"]
VISUALIZATION_CONFIG = CONFIG["visualization"]

print("Configuration loaded successfully!")
print("Available models:", list(MODEL_CONFIGS.keys()))
```

```python
# Set random seed for reproducibility
RANDOM_SEED = GENERAL_CONFIG["random_seed"]

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_SEED)

print("Random seed:", RANDOM_SEED)
```

```python
# Select computation device
if GENERAL_CONFIG["device"] == "auto":
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
else:
    device = torch.device(GENERAL_CONFIG["device"])

print("Available device:", device)
```

```python
print("Project:", GENERAL_CONFIG["project_name"])
print("Dataset:", GENERAL_CONFIG["dataset"])
print("Image shape:", GENERAL_CONFIG["image_shape"])
print("Input dimension:", GENERAL_CONFIG["input_dim"])
print("Batch size:", GENERAL_CONFIG["batch_size"])
print("Number of classes:", GENERAL_CONFIG["num_classes"])
```
